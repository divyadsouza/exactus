```python
import pandas as pd

pd.set_option('display.max_colwidth', None)
pd.set_option('display.max_columns', None)

def show_row(df, index=0):
    """Print every column of a single row with full text."""
    for col in df.columns:
        print(f'=== {col} ===')
        print(df.iloc[index][col])
        print()
```

## Original dataset — `train-00000-of-00001.parquet`

```python
original = pd.read_parquet('train-00000-of-00001.parquet')
print(f'{len(original)} rows, {len(original.columns)} columns')
original.head(3)
```

## NuExtract prompts — `nuextract_prompts.parquet`

```python
prompts = pd.read_parquet('nuextract_prompts.parquet')
print(f'{len(prompts)} rows, {len(prompts.columns)} columns: {list(prompts.columns)}')
prompts.head(3)
```

```python
# Change index to see a different row (0 to 5009)
show_row(prompts, index=2)
```
